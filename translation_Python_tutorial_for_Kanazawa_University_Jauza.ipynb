The four basic arithmetic operations on vectors are performed as shown below. NumPy generally performs calculations element-wise, meaning that operations are applied to each corresponding element individually.

```
import numpy as np

def main():
    na = np.array([1, 2, 3])
    nb = np.array([5, 7, 9])
    print(na + nb)
    print(na - nb)
    print(na * nb)
    print(nb / na)

main()
```

コピーをする際は気を使わなければならない点があります．あるベクトルから別のベクトル変数を生成，つまり，コピーでベクトルを生成した場合，その生成したベクトルを元のベクトルと別のものとして扱いたい場合は以下のようにしなければなりません．以下では，9行目と10行目で元のベクトルとコピーで生成されたベクトルの要素をそれぞれ別の値で変更していますが，それぞれ別の値にて要素が置換されていることがわかります．

There is an important point to keep in mind when copying vectors. When creating a new vector variable from an existing vector—that is, when creating a vector by copying another vector—you must follow the procedure shown below if you want to treat the newly created vector as an independent object, separate from the original vector. In the example below, the elements of the original vector and the copied vector are modified to different values on lines 9 and 10, respectively. As you can see, the corresponding elements are replaced with different values independently of each other.

```
import numpy as np

def main():
    na = np.array([1, 1])
    nb = na.copy()
    print(na, nb)
    na[0] = 2
    nb[0] = 3
    print(na, nb)

main()
```

一方で，以下のように = を使ってコピーをすると生成されたベクトルは元のベクトルの参照となってしまい，（この場合の）意図している操作は実現されません．上の挙動とこの挙動は把握していないと結構危険です．

On the other hand, if you copy a vector using the assignment operator (=), as shown below, the newly created variable becomes a reference to the original vector, rather than an independent copy. As a result, the intended operation (in this particular case) will not work as expected. Failing to understand the difference between the previous behavior and this behavior can be quite dangerous, as it may lead to unintended modifications and unexpected results.

```
import numpy as np

def main():
    na = np.array([1, 1])
    nb = na
    print(na, nb)
    na[0] = 2
    nb[0] = 5
    print(na, nb)

main()
```

### 2-6-4. 勾配法 (Gradient Descent Method)

<font color="blue">深層学習に関するニューラルネットワークやディープラーニングの詳細なアルゴリズムについては下記のサイトを参照して頂きつつ，ここでは，Pythonを用いた演算に焦点を当てて講義を進めます．</font>

深層学習の最も基本的な構成要素は行列の掛け算と微分です．PyTorch はこれを行うライブラリです．自動微分機能を提供します．微分をしたい変数の生成は以下のように行います．テンソル型の変数を生成する際に，requires_grad=True を追加するだけです．

<font color="yellow">For detailed explanations of neural networks and deep learning algorithms, please refer to the website linked below. In this lecture, however, we will focus on performing computations using Python.</font>

The most fundamental building blocks of deep learning are matrix multiplication and differentiation. PyTorch is a library designed to perform these operations. It provides an automatic differentiation feature. Variables for which we want to calculate derivatives can be created as shown below.When creating a tensor variable, we simply need to add requires_grad=True to enable automatic differentiation.

```
#!pip install torch
import torch

def main():
    tx = torch.tensor(5, dtype=torch.float, requires_grad=True)
    print(tx)

main()
```

ここで，以下の式を考えます．

$y = x^2 + 2$

これに対して以下の偏微分を考えることができます．

$\dfrac{\partial y}{\partial x} = 2x$

よって $x=5$ のときの偏微分の値は以下のように計算できます．

$\left.\dfrac{\partial y}{\partial x}\right|_{x=5}=10$

これを PyTroch で実装すると以下のようになります．微分は`backward()` によって行います．


Here, let us consider the following equation.

$y = x^2 + 2$

For this equation, we can consider the following partial derivative.

$\dfrac{\partial y}{\partial x} = 2x$

Therefore, when $x=5$, the value of the partial derivative can be calculated as follows.

$\left.\dfrac{\partial y}{\partial x}\right|_{x=5}=10$

Implementing this in PyTorch gives the following result. Differentiation is performed using `backward()`.


```
import torch

def main():
    tx = torch.tensor(5, dtype=torch.float, requires_grad=True)
    ty = tx**2 + 2
    ty.backward()
    print(tx.grad)

main()
```

なぜ微分を求めたいかというと，勾配降下法でパラメータをアップデートしたいからです．以下では勾配降下法を実装してみます．勾配降下法は関数の最適化法です．ある関数に対して極小値を計算するためのものです．以下のような手順で計算が進みます．


The reason we want to calculate derivatives is to update parameters using the gradient descent method. In the following section, we will implement gradient descent. Gradient descent is a method for optimizing functions. It is used to find a local minimum of a given function. The calculation proceeds through the following steps.

1.   初期変数$x_{t=0}$をランダムに用意します．
2.   もし変数$x_{t=0}$が最適値または，最適値に近いなら計算をやめます．
3.   パラメータを以下の式によって更新し，かつ，$t$ の値を $1$ だけ増やします．ここで，$\alpha$ は学習率と呼ばれる更新の大きさを決める値で，$g_t$ は $t$ のときの目的の関数の勾配です．<br>
    $x_{t+1}=x_t-\alpha g_t$
4.   ステップ2と3を繰り返します．

1. Randomly initialize the variable $x_{t=0}$.
2. If the variable $x_{t=0}$ is at or close to the optimal value, stop the calculation.
3. Update the parameter using the following equation and increment the value of $t$ by $1$. Here, $\alpha$ is called the learning rate, which determines the magnitude of the update, and $g_t$ is the gradient of the objective function at iteration $t$.

   $x_{t+1}=x_t-\alpha g_t$

4. Repeat steps 2 and 3.

ここでは具体的に，以下の関数を考えます．


$y=(x+1)^2+2$


この関数の極小値は$x=-1$で$y=2$ となります．
$x$の初期値を$x_1=1.6$とすると$y_1$は

  $y_1=(1.6+1)^2+2=8.76 \ne 2$

となり，極小値となっていません．

そのため，$y$を$x$に関して微分した勾配の逆方向に$x$を少しずつ動かしていきながら（更新しながら）$y$を最小化させていきます．

因みに，$x=-1$における勾配は

$\frac{dy}{dx}=g=2x+2＝2 \times -1+2=0$

となりますので，勾配$g$が$0$になるまで$x$を更新させていくと言い換えても良いかもしれません．


1.   Step1:$t=1$

$y_1$の$x_1$に関する微分は$2x_1+2$ですから，$x_1=1.6$における勾配は


$g_1=2x_1+2=2 \times 1.6 +2 =5.2$

となります．そして，$\alpha=0.1$とすると

$x_2=x_1-\alpha g_1=1.6-0.1\times5.2=1.08 \leq 1.6=x_1$


となり，少し$-1$に近づきましたね．ちなみに，

$y_2=(1.08+1)^2+2=6.326 \ne 2$

もう1ステップやってみましょう！

2.   Step2:$t=2$





$x_2=1.08$における勾配$g_2$は$4.16$となるので，

$x_3=x_2-\alpha g_2=1.08-0.1 \times 4.16=0.664 \leq 1.08=x_2$

となります．そして，

$y_3=(0.664+1)^2+2=4.76889 \ne 2$

ですね．

上記の操作を何度も反復させて，$x,y$を$-1,2$に近づける方法を最急降下法と呼びます．その際，$g=0$となれば反復をストップさせることになります．


Here, let us consider the following function.

$y=(x+1)^2+2$

This function reaches its minimum value of $y=2$ at $x=-1$.

If we set the initial value of $x$ to $x_1=1.6$, then $y_1$ becomes

$y_1=(1.6+1)^2+2=8.76 \ne 2$

Therefore, the function has not yet reached its minimum value.

To minimize $y$, we gradually move (update) $x$ in the direction opposite to the gradient obtained by differentiating $y$ with respect to $x$.

Incidentally, the gradient at $x=-1$ is

$\frac{dy}{dx}=g=2x+2=2 \times (-1)+2=0$

Therefore, we could also describe this process as repeatedly updating $x$ until the gradient $g$ becomes $0$.

1. Step 1: $t=1$

The derivative of $y_1$ with respect to $x_1$ is $2x_1+2$. Therefore, the gradient at $x_1=1.6$ is

$g_1=2x_1+2=2 \times 1.6+2=5.2$

Then, setting $\alpha=0.1$, we obtain

$x_2=x_1-\alpha g_1=1.6-0.1\times5.2=1.08 \leq 1.6=x_1$

Thus, $x$ has moved slightly closer to $-1$.

Incidentally,

$y_2=(1.08+1)^2+2=6.326 \ne 2$

Let's try one more step!

2. Step 2: $t=2$

Since the gradient $g_2$ at $x_2=1.08$ is $4.16$, we obtain

$x_3=x_2-\alpha g_2=1.08-0.1 \times 4.16=0.664 \leq 1.08=x_2$

Then,

$y_3=(0.664+1)^2+2=4.76889 \ne 2$

By repeatedly performing the above operations, we gradually bring $x$ and $y$ closer to $-1$ and $2$, respectively. This method is called the steepest descent method.

During this process, the iteration is stopped when $g=0$.


これを PyTorch を用いて実装すると以下のようになります．出力中，`Objective` は目的関数の値，`Solution` はその時点での解です．最終的に $x=-0.9968\simeq-1$ のとき，最適値 $y=2$ が出力されています．

Implementing this using PyTorch gives the following result. In the output, `Objective` represents the value of the objective function, while `Solution` represents the solution at the current iteration. Finally, when $x=-0.9968\simeq-1$, the optimal value $y=2$ is obtained.


```
import torch

def main():
    tx = torch.tensor(1.6, dtype=torch.float, requires_grad=True)
    epoch, update_value, lr = 1, 5, 0.1
    while abs(update_value) > 0.001:
        ty = (tx + 1)**2 + 2
        ty.backward()
        with torch.no_grad():
            update_value = lr * tx.grad
            tx = tx - update_value
            print("Epoch {:4d}:\tObjective = {:5.3f}\tSolution = {:7.4f}".format(epoch, ty, tx.numpy()))
            tx.requires_grad = True
        epoch = epoch + 1

main()
```

Principal Component Analysis (PCA) is one of the most widely used dimensionality reduction techniques and is a typical example of unsupervised learning. For example, PCA might be useful when analyzing the academic performance of 300 students across various subjects (Japanese language, social studies, mathematics, English, physics, physical education, classical literature, etc.) to help classify students into humanities-oriented, science-oriented, or sports-oriented groups.

In the Iris dataset, the four variables—sepal-length (sepal length), sepal-width (sepal width), petal-length (petal length), and petal-width (petal width)—are the primary focus of the analysis.

これらを次元圧縮するということは，4変数以下（例えば2変数くらいで）で元のデータを再現できるような状況を想定しています．

Reducing the dimensionality of these variables means aiming to represent the original data using fewer than four variables (for example, approximately two variables), while still retaining enough information to reconstruct the original data.

ここで重要となってくるのは、固有値及び累積寄与率です．細かい，解説は省きますが固有値は（主成分又はモードが）大きい順に並べられており，累積寄与率は，任意の数だけ固有値を上から足した数と固有値の総和との商で評価されます．

The important concepts here are eigenvalues and the cumulative explained variance ratio. Although we will omit the detailed explanation, the eigenvalues (corresponding to principal components or modes) are arranged in descending order, from largest to smallest. The cumulative explained variance ratio is calculated by dividing the sum of a specified number of the largest eigenvalues by the sum of all eigenvalues.

例えば，第2主成分くらいまでで累積寄与率が99程度であれば，元のデータを2変数で再現できているとみなします．

For example, if the cumulative explained variance ratio reaches approximately 99% by the second principal component, we can consider the original data to be adequately represented using just two variables.

さて，細かいことはおいておいて実際に解析していきましょう．

Now, let's set aside the finer details and proceed with the actual analysis.

```
dfs = df.loc[:, 'sepal_length':'petal_width'] # 特徴量だけを取り出す (Extract only the features.)
dfs.head()
```

```
# sklearnの標準化モジュールをインポート (Import the standardization module from scikit-learn)
from sklearn.preprocessing import StandardScaler

# データを変換する計算式を生成 (Generate the transformation parameters for standardizing the data)
sc = StandardScaler()
sc.fit(dfs)

# 実際にデータを変換 (Apply the transformation to the data)
z = sc.transform(dfs)
#print(z)

#主成分分析の実行 (Perform Principal Component Analysis (PCA))
pca = PCA()
pca.fit(z)

# データを主成分空間に写像 (Project the data into principal component space)
feature = pca.transform(z)
```

However, we will skip the rigorous mathematical explanation. I believe you will gain a more concrete understanding of this concept as you work through the exercises.

Now, let's examine the cumulative explained variance ratio. Since the cumulative explained variance ratio reaches approximately 0.99 (99%) by the second principal component, we can consider the original data to be adequately represented using the first two principal components.

```
# 累積寄与率を図示する (Visualize the cumulative explained variance ratio)
import matplotlib.ticker as ticker
plt.gca().get_xaxis().set_major_locator(ticker.MaxNLocator(integer=True))
plt.plot([0] + list( np.cumsum(pca.explained_variance_ratio_)), "-o")
plt.xlabel("Number of principal components")
plt.ylabel("Cumulative contribution rate")
plt.grid()
plt.show()
```

##3-8.Isomap

次元圧縮→PCA機能せず→サンプル数1万以下→[Isomap]

参照：http://neuro-educator.com/ml24/

Dimensionality Reduction → PCA Does Not Work Effectively → Number of Samples ≤ 10,000 → [Isomap]

Reference: http://neuro-educator.com/ml24/

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import preprocessing, decomposition, manifold
from sklearn import datasets

X,Y = datasets.make_moons(n_samples=200, noise=0.05, random_state=0)

sc=preprocessing.StandardScaler()
sc.fit(X)
X_norm=sc.transform(X)

isomap = manifold.Isomap(n_neighbors=10, n_components=2)
X_isomap = isomap.fit_transform(X)

%matplotlib inline
plt.figure(figsize=(10,10))
plt.subplot(3, 1, 1)
plt.scatter(X[:,0],X[:,1], c=Y)
plt.xlabel('x')
plt.ylabel('y')

plt.subplot(3, 1, 3)
plt.scatter(X_isomap[:,0],X_isomap[:,1], c=Y)
plt.xlabel('IM-1')
plt.ylabel('IM-2')

plt.show
```

#4.Scikit-LearnによるMLP・SVM・ランダムフォレストの実装 (Implementation of MLP, SVM, and Random Forest Using Scikit-Learn)

## 4-1. パッケージの読み込み (Importing Packages)

ここでは，Scikit-Learnから必要な道具だけ読み込んでいます．MLPClassifierは多層ニューラルネットワーク，train_test_splitは学習データを分割するための機能，StandardScalerはデータを標準化するための機能です．

Here, we import only the necessary tools from Scikit-Learn. `MLPClassifier` is used to implement a multilayer neural network, `train_test_split` is a function for splitting the dataset into training and testing subsets, and `StandardScaler` is a tool for standardizing the data.

```
import numpy as np
import pandas as pd
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV
import matplotlib.pyplot as plt
```

## 4-2. データの読み込み (Loading Data)

UCI machine learningrepositoryからアヤメのデータをダウンロードします．ここでのデータは150個のアヤメのサンプルに対して，それぞれsepal-length(外花被片の長さ)，sepal-width(外花被片の幅)，petal-length(内花被片の長さ)，petal-width(内花被片の幅)，speicies(アヤメの種類)の5変数を持っています（つまり150×5の行列ですね）．とはいえ，変数の名前が事前に付与されている訳はないので，df.columnsで変数名を定義しています．

We will download the Iris dataset from the UCI Machine Learning Repository. The dataset contains 150 iris flower samples, each with five variables: sepal-length (sepal length), sepal-width (sepal width), petal-length (petal length), petal-width (petal width), and species (iris species). In other words, the dataset can be represented as a 150 × 5 matrix. However, since the variable names are not assigned beforehand, we define the column names using `df.columns`.

```
df = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data', header=None) # web上のデータを読み込む (Load data from the web)
df.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species'] # 列名を付ける (Assign column names)
```

## 4-3. データの分解 (Data Decomposition)

データから特徴量と正解カテゴリーに分けます．特徴量とはニューラルネットワークに入力するデータで，具体的にはsepal-length(外花被片の長さ)，sepal-width(外花被片の幅)，petal-length(内花被片の長さ)，petal-width(内花被片の幅)のことです．一方，正解カテゴリーはspeicies(アヤメの種類)となります．

We separate the dataset into features and ground-truth class labels. Features are the data provided as inputs to the neural network. Specifically, these are sepal-length (sepal length), sepal-width (sepal width), petal-length (petal length), and petal-width (petal width). On the other hand, the ground-truth class label is species (the type of iris flower).

特徴量だけを取り出すために df.locコマンドを利用し，featuresという変数に格納します．

We use the `df.loc` command to extract only the features and store them in a variable called `features`.

```
features = df.loc[:, 'sepal_length':'petal_width'] # 特徴量だけを取り出す (Extract only the features)
display(features)
```

同様に，正解カテゴリーだけを取り出し，targetsという変数に格納します．

Similarly, we extract only the ground-truth class labels and store them in a variable called `targets`.

```
targets = df.loc[:, ['species']] # 正解カテゴリーを取り出す (Extract the ground-truth class labels)
display(targets)
```

## 4-4. Numpyへの変換と標準化 (Conversion to NumPy and Standardization)

下記のコマンドでは数学的な演算が可能となるようにNumpy形式にデータを変換しています．

The following command converts the data into NumPy format to enable mathematical operations.

```
# dataframeをndarrayに変換 (Convert the DataFrame into an ndarray)
#X = features.values
#y = targets.values
X = np.array(features)
#display(X)
y = np.array(targets)
y = y.ravel() # 配列の次元を落とす (Reduce the dimensionality of the array)
#display(y)
```

一般に統計解析を行う際にはデータを一旦，標準化（平均0，標準偏差1）させておくことが強く推奨されます．下記のコマンドで，標準化を実装しています．

In general, when performing statistical analysis, it is strongly recommended to standardize the data beforehand (mean = 0, standard deviation = 1). The following commands implement this standardization process.

```
scaler = StandardScaler() # scalerインスタンスの生成 (Create a scaler instance)
X = scaler.fit_transform(X) # 特徴量データの標準化 (Standardize the feature data)
display(X)
```

## 4-5. 訓練データと検証データ (Training Data and Validation Data)

用意されたデータを訓練データと検証データに分けるために，下記のように記述します．X_trainとy_trainで訓練させ，X_testとy_testで検証します．

To split the prepared dataset into training data and validation data, we write the code as shown below. The model is trained using `X_train` and `y_train`, and its performance is evaluated using `X_test` and `y_test`.

```
X_train, X_test, y_train, y_test = train_test_split(X, y) # 訓練データと検証データへの分割 (Split the data into training and validation sets)
```

## 4-6. モデルの生成と計算 (Model Creation and Computation)

### 4-6-1. MLP

さて，ニューラルネットワークを準備するために下記のようにコマンドを記述します．ここで，max_iter=10000は学習回数となりますが，十分に誤差が小さくなればmax_iter回以下でも計算を終了します．

Now, to set up the neural network, we write the commands as shown below. Here, `max_iter=10000` specifies the maximum number of training iterations. However, if the error becomes sufficiently small, the training process may terminate before reaching the maximum number of iterations.

```
clf = MLPClassifier(max_iter=10000) # 多層ニューラルネットワークの生成 (Create a multilayer neural network)
```

ここで，学習用データであるX_trainとy_trainを入力することで学習が開始されます．

Here, the training process begins when we provide `X_train` and `y_train`, which contain the training data, as inputs to the model.

ニューラルネットワークを利用する場合には下記のように記述します．

When using a neural network, write the code as shown below.

```
clf.fit(X_train, y_train)
```

さて，出力と正解カテゴリーの誤差がclf.loss_curve_に格納されているので可視化して確認してみましょう．学習回数が増えるほど誤差が小さくなっていることがわかります．

Now, since the error between the model's outputs and the correct class labels is stored in `clf.loss_curve_`, let's visualize it and examine the results. We can observe that the error decreases as the number of training iterations increases.

さて学習の成果を確認してみましょう！clf.predictコマンドを使うことで，予測結果を表示してくれます．

Now, let's check the results of the training! By using the `clf.predict` command, we can display the model's predictions.

```
clf.predict(X_test)
```

```
plt.plot(clf.loss_curve_)
plt.title("Loss curve") # グラフのタイトル (Graph title)
plt.xlabel("Iteration") # x軸は繰り返し数（エポック数） (The x-axis represents the number of iterations (epochs))
plt.ylabel("Loss") # y軸は損失値 (The y-axis represents the loss value)
plt.grid() # グリッド線を描く (Draw grid lines)
```

次に，clf.scoreコマンドを使うことで正解率を得ることになります．

Next, we can obtain the classification accuracy by using the `clf.score` method.

```
c = clf.score(X_test, y_test)
print("正解率は{}%です．(The accuracy is {}%.)".format(c * 100, c * 100)) # 検証データでの成績表示 (Display the performance on the validation data)
```

下記のソースコードを利用して，MLP,svm,ランダムフォレスト，ロジスティック回帰における正解率を比較せよ．

Using the source code provided below, compare the classification accuracies of MLP, SVM, Random Forest, and Logistic Regression.

```
import numpy as np
import pandas as pd

#from sklearn.neural_network import MLPClassifier #MLP
from sklearn import svm #svm
#from sklearn.ensemble import RandomForestClassifier #RF
#from sklearn.linear_model import LogisticRegression, LinearRegression #LR

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV
import matplotlib.pyplot as plt

df = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data', header=None) # web上のデータを読み込む (Load data from the web)
df.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species'] # 列名を付ける (Assign column names)

features = df.loc[:, 'sepal_length':'petal_width'] # 特徴量だけを取り出す (Extract only the features)
#display(features)

targets = df.loc[:, ['species']] # 正解カテゴリーを取り出す (Extract the ground-truth class labels)
#display(targets)

# dataframeをndarrayに変換 (Convert the DataFrame into an ndarray)
#X = features.values
#y = targets.values
X = np.array(features)
#display(X)
y = np.array(targets)
y = y.ravel() # 配列の次元を落とす (Flatten the array into one dimension)
#display(y)

scaler = StandardScaler() # scalerインスタンスの生成 (Create a scaler instance)
X = scaler.fit_transform(X) # 特徴量データの標準化 (Standardize the feature data)
#display(X)

X_train, X_test, y_train, y_test = train_test_split(X, y) # 訓練データと検証データへの分割 (Split the data into training and validation sets)

#clf = MLPClassifier(max_iter=10000) # 多層ニューラルネットワークの生成 (Create a multilayer neural network)
clf = svm.SVC(max_iter=10000) #svm
#clf = RandomForestClassifier(random_state=1234) #RF
#clf = LogisticRegression(max_iter=10000) #LR

clf.fit(X_train, y_train)

clf.predict(X_test)

c = clf.score(X_test, y_test)
print("正解率は{}%です．(The accuracy is {}%.)".format(c * 100, c * 100)) # 検証データでの成績表示 (Display performance on the validation data)
```
